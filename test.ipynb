# Modélisation Prédictive

## Définition du Problème

En complément de l’analyse exploratoire, ce projet intègre une composante de modélisation prédictive visant à anticiper les comportements d’achat des clients.

Deux tâches d’apprentissage supervisé sont définies, chacune répondant à un objectif business précis.

---

## Tâche 1 — Prédiction du Reorder (Classification)

### Objectif

Prédire si un produit sera réacheté par un client lors d’une commande donnée.

### Variable Cible

`reordered`
Variable binaire :

* 1 → le produit a déjà été acheté auparavant par le client
* 0 → il s’agit d’un premier achat

### Unité d’Observation

Chaque observation correspond au triplet :

```
(user_id, product_id, order_id)
```

Chaque ligne représente un produit spécifique dans une commande spécifique pour un utilisateur donné.

### Justification Business

La prédiction du réachat permet :

* D’améliorer les recommandations personnalisées
* D’optimiser les campagnes marketing ciblées
* D’augmenter la fidélisation client
* D’optimiser la mise en avant des produits

Ce modèle contribue directement à l’augmentation du revenu par anticipation des comportements d’achat.

---

## Tâche 2 — Prédiction de la Taille du Panier (Régression)

### Objectif

Prédire le nombre total de produits qu’un client inclura dans une commande.

### Variable Cible

`cart_size`
Nombre total d’articles présents dans une commande.

### Unité d’Observation

Chaque observation correspond à :

```
order_id
```

Chaque ligne représente une commande complète.

### Justification Business

La prédiction de la taille du panier permet :

* D’anticiper le volume de commandes
* D’optimiser la gestion des stocks
* D’améliorer la planification logistique
* De réduire les ruptures et les surstocks

Ce modèle soutient l’efficacité opérationnelle et la performance logistique.

---

## Cadre Méthodologique

La construction des modèles suit une démarche structurée et reproductible.

### 1. Construction du Dataset

* Fusion des tables nécessaires
* Définition claire des variables cibles
* Vérification de la cohérence des données
* Prévention du data leakage

### 2. Séparation Entraînement / Test

* Division du dataset en ensembles d’apprentissage et de test
* Garantie d’une évaluation objective
* Respect de la cohérence temporelle si nécessaire

### 3. Feature Engineering

Création de variables explicatives pertinentes :

* Variables comportementales (fréquence d’achat, taux de reorder)
* Variables temporelles (récence, jour, heure)
* Agrégations au niveau produit
* Agrégations au niveau utilisateur

### 4. Développement des Modèles

Pour chaque tâche :

* Un modèle baseline (référence interprétable)
* Un modèle plus avancé visant l’amélioration des performances

### 5. Évaluation

**Pour la classification :**

* Precision
* Recall
* F1-score
* ROC-AUC

**Pour la régression :**

* MAE
* RMSE
* R²

Les performances seront comparées de manière cohérente entre les modèles.

---

## Conclusion

Les deux tâches prédictives répondent à des enjeux complémentaires :

* La prédiction du reorder améliore la personnalisation et la fidélisation.
* La prédiction de la taille du panier renforce l’anticipation logistique et la gestion des stocks.

Cette approche démontre la capacité de la modélisation à transformer des données transactionnelles en leviers décisionnels concrets.
